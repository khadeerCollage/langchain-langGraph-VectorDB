```python
# tools 

from langchain_community.tools import ArxivQueryRun,WikipediaQueryRun
from langchain_community.utilities import ArxivAPIWrapper,WikipediaAPIWrapper
```

```python
api_wrapper_arxiv = ArxivAPIWrapper(top_k_results=2,doc_content_chars_max=500)
arxiv  = ArxivQueryRun(api_wrapper=api_wrapper_arxiv, description="Query arxiv papers")
print(arxiv.name) 
```

```python
arxiv.invoke("Graph Neural Networks")
```

```python
api_wrapper_wiki = WikipediaAPIWrapper(top_k_results=2,doc_content_chars_max=1000)
wiki  = WikipediaQueryRun(api_wrapper=api_wrapper_wiki, description="Query wikipedia articles")
print(wiki.name) 
```

```python
wiki.invoke("Graph Neural Networks")
```

```python
from dotenv import load_dotenv
load_dotenv()

import os

os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
### Taviley search tool

from langchain_community.tools.tavily_search import TavilySearchResults

tavily = TavilySearchResults()
```

```python
tavily.invoke("Graph Neural Networks")
```

```python
## combine all these tools in the list

tools = [arxiv, wiki, tavily]
```

```python
## intialize the LLM model


# profile = {
#     "max_input_tokens": 131072,
#     "max_output_tokens": 32768,
#     "image_inputs": True,
#     "audio_inputs": True,
#     "video_inputs": True,
#     "tool_calling": True
# }

# # create the client with media-enabled profile
# llm = ChatGroq(profile=profile, model="llama-3.3-70b-versatile")

from langchain_groq import ChatGroq

llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
llm.invoke("Explain graph neural networks in simple terms.")
```

```python
llm_with_bind  = llm.bind_tools(tools=tools)
```

```python
llm_with_bind.invoke("Explain graph neural networks in simple terms.")
```
